```python
class Pokemon:
    type_weaknesses = {
             "Fire": ["Grass"],
             "Grass": ["Fire"]
         }
    def __init__(self, name, type, level, health):
        self.name = name
        self.ptype = type
        self.level = level
        self.health = health

    def level_up(self):
        self.level += 1
        self.health += 10
        print(f"{self.name} leveled up to {self.level}!")

    def calculate_damage(self, target):
             damage = self.level * 1.5  # Base damage
             print(f"Base damage: {damage}")

             if target.ptype in self.type_weaknesses.get(self.ptype, []):
                 damage *= 2  # Double damage if weakness
                 print(f"Damage after weakness: {damage}")
             else:
                 print("Weakness not found")  # Add this for debugging

             return damage

    def attack(self, target):
         damage = self.calculate_damage(target)
         target.take_damage(damage)
         print(f"{self.name} attacks {target.name}!")

    def take_damage(self, damage):
         print(f"{self.name} health before damage: {self.health}")
         self.health -= damage
         print(f"{self.name} takes {damage} damage! HP left: {self.health}")
```

```python
class Trainer:
    def __init__(self, name):
        self.name = name
        self.pokedex = PokeDex()

    def catch_pokemon(self, pokemon):
        self.pokedex.catch_pokemon(pokemon)

    def train_pokemon(self, pokemon_name):
        for pokemon in self.pokedex.pokemon_list:
            if pokemon.name == pokemon_name:
                print(f"trained {pokemon.name}!")
                pokemon.level_up()
                return
        print(f"{pokemon_name} is not in the PokeDex.")

    def show_pokedex(self):
        print(f"{self.name}'s Pokedex:")
        self.pokedex.show_pokedex()
```

```python
class PokeDex:
    def __init__(self):
        self.pokemon_list = []

    def catch_pokemon(self, pokemon):
        self.pokemon_list.append(pokemon)
        print(f"{pokemon.name} caught {pokemon.name}!")

    def release_pokemon(self, pokemon_name):
        for pokemon in self.pokemon_list:
            if pokemon.name == pokemon_name:
                self.pokemon_list.remove(pokemon)
                print(f"released {pokemon.name}.")
                return
        print(f"{pokemon_name} is not in PokeDex.")

    def show_pokedex(self):
        if not self.pokemon_list:
            print("No Pokemon in the PokeDex yet.")
        else:
            for pokemon in self.pokemon_list:
                print(f"- {pokemon.name} ({pokemon.ptype})")
```

```python
bulbasaur = Pokemon('Bulbasaur', 'Grass', 10, 100)
charizard = Pokemon('Charizard', 'Fire', 15, 150)

satoshi = Trainer('Satoshi')
ashley = Trainer('Ashley')
ashley.catch_pokemon(bulbasaur)
satoshi.catch_pokemon(charizard)
ashley.show_pokedex()
satoshi.show_pokedex()
charizard.attack(bulbasaur)
```
